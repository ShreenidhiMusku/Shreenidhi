```python
import numpy as np
student1 = np.array([85, 78, 90])
student2 = np.array([72, 88, 80])
student3 = np.array([95, 92, 87])
print("L2 Norm of Student 1:", np.linalg.norm(student1))
print("L2 Norm of Student 2:", np.linalg.norm(student2))
print("L2 Norm of Student 3:", np.linalg.norm(student3))
print("Student 1 & Student 2:",
      np.dot(student1, student2) /
      (np.linalg.norm(student1) * np.linalg.norm(student2)))
print("Student 1 & Student 3:",
      np.dot(student1, student3) /
      (np.linalg.norm(student1) * np.linalg.norm(student3)))
print("Student 2 & Student 3:",
      np.dot(student2, student3) /
      (np.linalg.norm(student2) * np.linalg.norm(student3)))

# This cell (i0W2q0xBAz93) executed successfully without errors.
# A 'LinAlgError: Singular matrix' was observed in cell _oCM2DF6FHnw.
```
